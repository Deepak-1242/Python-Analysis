# Movie Recommendation System 

## `Getting The data`

1. Data Collection
The dataset used for this project was sourced from a publicly available TMDB movie database from kaggle. It included essential movie attributes such as title, genre, cast, director, and plot description.

## Importing the dependencies

```python
import pandas as pd 
import numpy as np 
import ast
```

```python
movies = pd.read_csv("./CSV-Files/tmdb_5000_movies.csv")
credits = pd.read_csv("./CSV-Files/tmdb_5000_credits.csv")
```

```python
movies
```

```python
movies = movies.merge(credits,on="title")
movies.columns
```

## `Preparing the Data and Cleaning Data`

    1. Data Preprocessing
T1 computation.

```python
movies =movies[["movie_id","title","genres","keywords","overview","cast","crew"]]
```

```python
movies.dropna(inplace=True)
movies.isna().sum()
movies.drop_duplicates(inplace=True)
```

```python
movies.reset_index(inplace=True,drop=True)
```

```python
movies.head()
```

```python
def convert_genre(obj):
    genre_list = []
    for i in ast.literal_eval(obj):
            genre_list.append(i["name"])
    return genre_list
    

def convert_keyword(obj):
    keyword_list = []
    for i in ast.literal_eval(obj):
            keyword_list.append(i["name"])
    return keyword_list


def convert_cast(obj):
    cast_list = []
    counter =0 
    for i in ast.literal_eval(obj):
        if counter!=3:
            cast_list.append(i["name"])
            counter = counter +1
        else:
            break

    return cast_list
```

```python
movies['cast']=movies.cast.apply(convert_cast)
```

```python
movies['genres'] = movies.genres.apply(convert_genre)
```

```python
movies.keywords = movies.keywords.apply(convert_keyword)
```

```python
movies.overview = movies.overview.apply(lambda x:x.split())
```

```python
def convert_crew(obj):
    crew_list = []
    for i in ast.literal_eval(obj):
        if i['job'] == "Director":
            crew_list.append(i["name"])
            break

    return crew_list
```

```python
movies.crew=movies.crew.apply(convert_crew)
```

```python
movies.genres=movies.genres.apply(lambda x:[i.replace(" ","") for i in x])
movies.keywords=movies.keywords.apply(lambda x:[i.replace(" ","") for i in x])
movies.cast=movies.cast.apply(lambda x:[i.replace(" ","") for i in x])
movies.crew=movies.crew.apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.overview = movies.overview
```

```python
movies.head()
```

```python
movies['tags'] = movies.overview + movies.genres + movies.keywords + movies.cast+ movies.crew
```

```python
movies.drop(columns=['genres', 'keywords', 'overview', 'cast', 'crew'], axis=1,inplace=True)                                
```

```python
movies.head()
```

```python
movies.tags=movies.tags.apply(lambda x:" ".join(x))
movies.tags=movies.tags.apply(lambda x:x.lower())
```

```python
movies.head()
```

## Preparing the  data for modelling 

```python
movies.tags[0],movies.tags[1]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words="english")
```

```python
vectors = cv.fit_transform(movies.tags).toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out()[0:100]
```

```python
import nltk 
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
def stem(text):
    y =[]
    for i in text.split():
        y.append(ps.stem(i))
    return ' '.join(y)
```

```python
movies.tags = movies.tags.apply(stem)
```

```python
movies.head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
similarity
```

```python
similarity[1]
```

```python
similarity[0]
```

```python
list(enumerate(similarity[1]))
```

```python
def recommend(movie):
    movies_index = movies[movies.title == movie].index[0]
    distance = similarity[movies_index]
    recommended_movies = sorted(list(enumerate(distance)),reverse=True,key=lambda x:x[1])[1:10]

    for i in recommended_movies:
        print(movies.iloc[i[0]].title)
```

```python
recommend('Avatar')
```

```python
for i in movies.title.head():
    print(f"Movie Name: {i}")
    print("Recommended Movies:")
    recommend(i)
    print("*************************************************************")
```
