```python
import streamlit as st
import pandas as pd 
import joblib

movies_dict = joblib.load('./movies_dict.pkl')

movies = pd.DataFrame(movies_dict)

def recommend(movie):
    movies_index = movies[movies.title == movie].index[0]
    distance = similarity[movies_index]
    recommended_movies = sorted(list(enumerate(distance)),reverse=True,key=lambda x:x[1])[1:10]

    movies_list = []
    for i in recommended_movies:
        movies_list.append(movies.iloc[i[0]].title)
    return movies_list

similarity = joblib.load('./similarity.pkl')

st.header("Movie Recommendation System")

selected_option = st.selectbox("Enter the movie name: ", movies.title.values)

if st.button("Recommend"):
       recommended_movie = recommend(selected_option)
       st.write(recommended_movie)
```
